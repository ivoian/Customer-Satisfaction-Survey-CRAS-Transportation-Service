```python
import matplotlib.pyplot as plt
import numpy as np

questions = [
    "Improvements in transportation services",
    "Ongoing performance evaluation",
    "Effective online scheduling system",
    "Issues with transportation services",
    "Convenient service schedules",
    "Clarity of staff explanations",
    "Satisfaction with waiting time",
    "Accessibility of the service team",
    "Perception of a welcoming environment"
]

yes = np.array([3, 0, 98, 5, 0, 92, 37, 38, 43])
no = np.array([97, 100, 2, 95, 100, 8, 63, 62, 57])

x = np.arange(len(questions))

width = 0.4

fig, ax = plt.subplots(figsize=(14, 8))
bars1 = ax.bar(x - width/2, yes, width=width, label='Yes', color='#4a90e2')
bars2 = ax.bar(x + width/2, no, width=width, label='No', color='#e94e77')

ax.set_ylabel('Percentage (%)')
ax.set_title('Survey Responses on CRAS Transportation Service')
ax.set_xticks(x)
ax.set_xticklabels(questions, rotation=45, ha='right')
ax.legend()

for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.0f}%',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # offset relative to top
                    textcoords="offset points",
                    ha='center', va='bottom')

plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
